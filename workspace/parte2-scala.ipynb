## Ejercicio 4 — Función para determinar si una mano se pasa de 21

Comparamos la puntuación con 21 mediante bust. Probamos 18, 21, 22 y 30 y obtenemos false, false, true y true, sin modificar variables externas.

```scala
def bust (puntuacion:Int): Boolean = {
    if (puntuacion > 21) {
        return(true)
    }
    else {
        return(false)
    }
}

println(bust(18))
println(bust(21))
println(bust(22))
println(bust(30))
```

```scala
def maxHand (mano1:Int, mano2:Int): Int = {
    if (mano1 > mano2){
        return mano1
    }
    else {
        return mano2
    }
}

println(maxHand(17,19))
println(maxHand(20,18))
println(maxHand(21,21))
```

## Ejercicio 6 — Decidir el ganador de una partida

Comprobamos qué manos superan 21 con bust y elegimos el resultado según las condiciones. Probamos los cinco casos y anotamos la rama utilizada.

```scala
def ganador (handA:Int, handB:Int): Int = {
    if ((bust(handA) && bust(handB)) == true){
        return 0
    }
    else if (bust(handA) == true && bust(handB) == false){
        return handB
    }
    else if (bust(handA) == false && bust(handB) == true){
        return handA
    }
    else {
        maxHand(handA,handB)
    }
}

println(ganador(26,20))
println(ganador(18,22))
println(ganador(24,25))
println(ganador(17,19))
println(ganador(21,20))
```

- 26 20: El primer else if
- 18 22: El segundo else if
- 24 25: if
- 17 19: else
- 21 20: else

## Ejercicio 9 — Recorrer un Array con while

Recorremos las cinco posiciones con un contador y un while. Mostramos cada puntuación y el resultado de bust, e incrementamos el índice en cada vuelta.

```scala
val manos = Array(17, 24, 21, 19, 26)
var i = 0
while (i < manos.length){
    print(manos(i))
    println(" " +bust(manos(i)))
    i += 1
}
```

### ¿Por qué el contador debe ser mutable?
Porque si no, no podremos incrementar la variable para recorrer la longitud del array de forma correcta.
### ¿Qué ocurriría si no incrementaras i?
Que haríamos un bucle infinito y el ejercicio no tendría sentido.
### ¿Qué elementos del ejercicio corresponden al estilo imperativo?
El bucle while, el contador mutable i y su incremento con i += 1, porque indicamos paso a paso como recorrer el array

## Ejercicio 10 — Listas e inmutabilidad

Añadimos un nombre al principio con :: y obtenemos una lista nueva. Mostramos ambas listas, la longitud de la nueva lista y su orden invertido.

```scala
val jugadores = List("Alex", "Chen", "Marta")
println(jugadores.mkString(", "))

val newList = "Sindhu"::(jugadores)

println("A continuación mostramos las dos listas:")
println(jugadores.mkString(", "))
println(newList.mkString(", "))

println("Longitud de newList: " +newList.length)
println("Lista invertida: " +newList.reverse.mkString(", "))
```

### Explica por qué la operación con :: produce una lista nueva en lugar de modificar la lista original:
Porque la lista original no se puede modificar porque es inmutable, por eso el operador "::" crea una lista nueva.

## Ejercicio 13 — foreach y funciones como valores

Evaluamos las mismas manos con while y foreach. Obtenemos las mismas salidas y comparamos el uso del contador en las dos versiones.

```scala
val manos = Array(17, 24, 21, 26, 18)
var i = 0

while(i < manos.length){
    println("¿Es mayor " +manos(i) +" que 21? " +bust(manos(i)))
    i += 1
}

println(" ")

manos.foreach { manos =>
    println("¿Es mayor " +manos +" que 21? " +bust(manos))
}

```

### ¿Cuál necesita un contador?
El while

### ¿Cuál necesita una variable var para recorrer la colección?
El while ya que necesita una variable que pueda cambiar de valor para recorrer el array

### ¿Cuál se aproxima más al estilo funcional presentado en el material del curso (Datacamp)?
El foreach porque pasa una función que se aplica a cada elemento, sin necesitar un contador mutable

## Ejercicio 15 — Programa integrado: torneo de Twenty-One

Relacionamos los nombres de la lista con las puntuaciones de cada ronda. Recorremos las manos, comprobamos cuáles superan 21 y calculamos la mejor puntuación válida. Reiniciamos el contador antes de recorrer la segunda ronda.

```scala
val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)
val manosRonda2 = Array(22, 19, 20, 21)
var i = 0


def mejorPuntuacion (puntuaciones: Array[Int]): Int = {
    var bandera = 0
    var i = 0
    while(i < puntuaciones.length) {
        if (bandera < puntuaciones(i) && !bust(puntuaciones(i))){
            bandera = puntuaciones(i)
        }
        i += 1
    }
    return bandera
}

while (i < jugadores.length){
    println(jugadores(i) +" con la siguiente puntuación: (" +manos(i) +") se ha pasado de 21? " +bust(manos(i)))
    i += 1
}

println("La mejor puntuacion es: " +mejorPuntuacion(manos))

println("\nEmpezamos la segunda ronda:")

i=0

while (i < jugadores.length){
    println(jugadores(i) +" con la siguiente puntuación: (" +manosRonda2(i) +") se ha pasado de 21? " +bust(manosRonda2(i)))
    i += 1
}

println("La mejor puntuacion es: " +mejorPuntuacion(manosRonda2))

```

### ¿Qué partes de tu solución son mutables?
Los contadores i, bandera y los arrays manos y manosRonda2 son mutables, aunque en este caso no modificamos los elementos de los arrays.

### ¿Qué partes son inmutables?
La lista jugadores

### ¿Dónde utilizas funciones?
Utilizamos bust para comprobar si se pasa de 21 y mejorPuntuacion para obtener la puntuación válida más alta

### ¿Dónde utilizas estructuras de control?
Utilizamos while para recorrer los jugadores y las puntuaciones, e if para actualizar la mejor puntuación cuando corresponde

### ¿Qué parte de tu solución consideras más cercana al estilo funcional?
El uso de una lista inmutable y de funciones que reciben datos y devuelven resultados, como bust

### ¿Qué parte consideras más cercana al estilo imperativo?
Los bucles while y las variables que vamos actualizando, como los contadores y bandera.
